# VeReMi-Graph Demo Explorer
Load and inspect one demo temporal graph.


```python
from pathlib import Path
import json
import pandas as pd
import torch
DEMO_ROOT = Path.cwd().parent
CATEGORY = 'medium'
print(DEMO_ROOT)
```

```python
nodes = pd.read_parquet(DEMO_ROOT/'data'/CATEGORY/'nodes.parquet')
edges = pd.read_parquet(DEMO_ROOT/'data'/CATEGORY/'edges.parquet')
snapshot = pd.read_parquet(DEMO_ROOT/'data'/CATEGORY/'snapshot.parquet')
display(nodes.head())
display(edges.head())
display(snapshot)
```

```python
graph = torch.load(DEMO_ROOT/'pyg'/f'{CATEGORY}.pt', weights_only=False)
print(graph)
print('x:', tuple(graph.x.shape))
print('edge_index:', tuple(graph.edge_index.shape))
print('edge_attr:', tuple(graph.edge_attr.shape))
print('labels:', tuple(graph.edge_y_multiclass.shape))
print('graph_context:', tuple(graph.graph_context.shape))
```
